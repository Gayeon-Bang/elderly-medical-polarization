```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings 
warnings.filterwarnings('ignore')
plt.rcParams['font.family']= 'NanumGothic' 
plt.rcParams['axes.unicode_minus']=False
# 데이터 정리
categories = [
    "살거리", "할거리", "놀거리", "먹거리", "쉴거리", "볼거리",
    "현지인/소통", "물가", "청결/위생", "교통편의성", "편의시설", "안전/치안", "상도덕"
]
domestic = [27, 22, 18, 16, 12, 10, 7, 52, 39, 30, 29, 6, 29]
overseas = [16, 15, 18, 16, 11, 8, 39, 20, 24, 21, 22, 19, 15]
difference = [d - o for d, o in zip(domestic, overseas)]

# 시각화
plt.figure(figsize=(12, 6))
plt.plot(categories, domestic, marker='o', label='국내 여행', color='#ff4d73')
plt.plot(categories, overseas, marker='o', label='해외 여행', color='#3478f6')

# 차이 강조
for i, diff in enumerate(difference):
    plt.text(i, max(domestic[i], overseas[i]) + 2, f'{diff:+}', 
             ha='center', fontsize=9, color='gray')

plt.title('[국내 vs 해외 여행] 불만족 요인 비교', fontsize=14, fontweight='bold')
plt.xlabel('평가 요인')
plt.ylabel('응답률 (%)')
plt.legend()
plt.xticks(rotation=45)
plt.grid(True, linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```
